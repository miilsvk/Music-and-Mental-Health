### 3. Преобразование типов данных

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

num_col = [
    'Age',
    'Hours per day',
    'BPM',
    'Anxiety',
    'Depression',
    'Insomnia',
    'OCD'
]

for col in num_col:
    df[col] = pd.to_numeric(df[col])
```
